## Task 3: Weapon Class:


```python
from Dice import Dice
from items import Weapon
from Character import Character
```

```python
# Attack with a weapon and check the damage taken by Enemy
sword = Weapon("Sword", Dice(6))
hero = Character("Hero", weapon=sword)
enemy = Character("Enemy")

hero.attack(enemy)

print(enemy.hp)
```

```python
# Test attacking without an equipped weapon
hero = Character("Hero")
enemy = Character("Enemy")

hero.attack(enemy)

print(enemy.hp)
```

## Discussion and Conclusion
We have successfully made the weapon class to use the Dice to calculate random damage. Based on oure test we can confirme a character equipped with a weapon deals damage based on a dice roll, while a character without a weapon uses the deafult damage of 10.
