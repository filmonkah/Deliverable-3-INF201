## Task 4: Character specialization 

```python
from Character import Character, Mage, Warrior
from Dice import Dice
from items import Weapon
```

```python
#here I`m creating a warrior and showing it`s attributes 
warrior = Warrior("Geir")
print(repr(warrior)) #shows name, Hp, damage and armor 
print(f"Is a Character: {isinstance(warrior, Character)}")

#testing that the armor reduces incoming damage 
warrior.take_damage(11) #11-5 armor = 8 damage
print(f"After 11 damage: {warrior.hp} HP")
warrior.take_damage(4)#4 - 5 armor is below 0, so no damage 
print(f"After 4 damage: {warrior.hp} HP")
warrior.take_damage(200) #more damage than the warrior has HP
print(f"After 200 damage: {warrior.hp} HP, alive: {warrior.is_alive}")
```

```python
#mage without weapon 
#creating a mage and an enemy, and letting the mage attack five times 
mage = Mage("Kari")
enemy = Character("Enemy")
print(repr(mage))

for attack_number in range(1, 6): #five attacks in a row
    mage.attack(enemy) #spell if mane >= 10, otherwise base attack 
    print(f"Attck {attack_number}: enemy HP = {enemy.hp}, mage mana = {mage.mana}")
```

```python
#mage with weapon 
#testing that spells ignore the weapon, but base attacks use it 
staff = Weapon("Staff", Dice(6)) # weapon that deals 1-6 damage 
mage = Mage("kari", weapon=staff)
enemy = Character("Enemy") # new enemy eith 100HP

mage.attack(enemy) #the first attack is a spell 
assert enemy.hp == 80 # spell deals double damage 2*10, not weapon damage 
mage.attack(enemy) #second spell 
mage.attack(enemy) #third spell, mana = 0 
print(f"After 3 spells: enemy HP = {enemy.hp}, mage = {mage.mana}")

hp_before = enemy.hp #remember HP before the base attack 
mage.attack(enemy) #not enough mana so base attack with the staff 
Weapon_damage = hp_before - enemy.hp 
assert 1 <= Weapon_damage <= 6
print(f"Base attack with staff dealt {Weapon_damage} damage, mana = {mage.mana}")
```

## Results and discussion 
The warrior`s armor redces 11 damage to 6 (94 HP), and 4 damage did nothing, since max() stops the armor from healing. after 200 damage it had 0 HPand was not alive. 

The mage cast three spells of 20 damage, then used base attacks that restored 5 mana each, and after five attacks the enemy had 20 HP left. Wuth a stadd, spells still dealt 20 damage, while the base attack dealt 3, so spells ignore the weapon.

## Taks 5: Running battles 


```python
from battle import Battle 
from Character import Character, Mage, Warrior
from Dice import Dice
from items import Weapon
```

```python
# Test play_round() on two normal characters 
battle = Battle(Character("Hero"), Character("Enemy")) # two characters without weapons 
print(battle.play_round()) #first round 
print(battle.play_round()) # second round 
```

```python
#Full battle between two normal characteres without weapons 
battle = Battle(Character("Hero"), Character("Enemy")) # a new battle with full HP
winner = battle.simulate_battle() #play until one dies 
for line in battle.log: #Show every round 
    print(line)
print(f"Winner: {winner.name} after {battle.round_number} rounds")
```

```python
# Warrior against Mage without weapons 
battle = Battle(Warrior("Geir"), Mage("Kari")) #warrior attacks first 
winner = battle.simulate_battle()
for line in battle.log: 
    print(line)
print (f"Winner: {winner.name} after {battle.round_number} rounds")
```

```python
#Warrior with a sword against Mage with a staff 
sword = Weapon("Sword", Dice(12)) #deals 1-12 damage 
staff = Weapon("Staff", Dice(6)) #deals 1-6 damage 
battle = Battle(Warrior("Geir", weapon=sword), Mage("Kari", weapon=staff)) #both have weapons 
winner = battle.simulate_battle()
for line in battle.log:
    print(line)
print(f"Winner: {winner.name} after {battle.round_number} rounds")
```

```python
#Simulate 100 battles ro see who wins most often with weapons
wins = {"Geir": 0, "Kari": 0}
for battle_number in range(100):
    warrior = Warrior("Geir", weapon=Weapon("Sword", Dice(12))) # new warrior with full health 
    mage = Mage("Kari", weapon=Weapon("Staff", Dice(6))) #new mage with full health and mana 
    winner = Battle(warrior, mage). simulate_battle()
    wins[winner.name] += 1 
print(wins)
```

```python
# To warriors whose damage is lower than the armor can never hurt eachother 
battle = Battle(Warrior("Weak A", damage=3), Warrior("Weak B", damage=3)) #3 - 5 armor = 0 damage 
winner = battle.simulate_battle() #stops after MAX_ROUNDS
print(f"Winner: {winner}, rounds played: {battle.round_number}")
```

## Results and discussion
Two normal characters lost 10 HP each round, and the hero won after 10 rounds with 10 HP left, because it attacks first. Warrior against Mage also lasted 10 rounds, and the warrior won with 5 HP left. With weapons the damage was random: in one battle the warrior won after 15 rounds, and in 100 battles the warrior won 37 times and the mage 63 times. MAX_ROUNDS stops battles that can never end, such as two warriors whose damage is lower than the armor.
